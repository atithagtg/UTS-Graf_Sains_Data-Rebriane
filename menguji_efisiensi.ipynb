```python
import time
import scipy.sparse as sp
from pathlib import Path
```

```python
# Jika variabel belum tersedia, muat dari file
DATA_DIR = Path.cwd() / "data" / "processed"

if "vsm_user" not in globals():
    vsm_user = sp.load_npz(DATA_DIR / "VSM_user_P1.npz")

if "vsm_item" not in globals():
    vsm_item = sp.load_npz(DATA_DIR / "VSM_item_P1.npz")
```

```python
# Ukuran penyimpanan aktual matriks CSR:
# data + indices + indptr
def ukuran_memori_csr(matriks):
    return (
        matriks.data.nbytes
        + matriks.indices.nbytes
        + matriks.indptr.nbytes
    )

def waktu_transpose(matriks, pengulangan=5):
    mulai = time.perf_counter()

    for _ in range(pengulangan):
        hasil = matriks.transpose().tocsr()

    selesai = time.perf_counter()
    return (selesai - mulai) / pengulangan

hasil_benchmark = {
    "VSM User-Item": vsm_user,
    "VSM Item-User": vsm_item
}

for nama, matriks in hasil_benchmark.items():
    print(f"\n{nama}")
    print("Shape:", matriks.shape)
    print("Nonzero:", matriks.nnz)
    print(
        "Memori CSR (MB):",
        round(ukuran_memori_csr(matriks) / (1024 ** 2), 3)
    )

print("\nRata-rata waktu transpose CSR:")
print(
    "User-Item -> Item-User:",
    round(waktu_transpose(vsm_user), 6),
    "detik"
)
print(
    "Item-User -> User-Item:",
    round(waktu_transpose(vsm_item), 6),
    "detik"
)
```
