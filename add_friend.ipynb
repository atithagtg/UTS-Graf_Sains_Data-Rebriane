2. Untuk mengakses database lain dalam satu instance sebenarnya bisa dilakukan melalui CYPHER, seperti menggunakan 
- klausa **USE**: 

USE nama_database -> menentukan database yang menjadi target query

MATCH (n) -> mengambil node dari database yang dipilih

RETURN n -> menampilkan node yang ditemukan

LIMIT 10;  -> membatasi hasil yang ditampilkan

```python
def add_friend(driver, name, friend_name):
    with driver.session(database="neo4j") as session:
        session.run(
            """
            MERGE (a:Person {name: $name})
            MERGE (friend:Person {name: $friend_name})
            MERGE (a)-[:KNOWS]->(friend)
            """,
            name=name,
            friend_name=friend_name
        )
```

```python
def print_friends(driver, name):
    with driver.session(database="neo4j") as session:
        result = session.run(
            """
            MATCH (a:Person)-[:KNOWS]->(friend)
            WHERE a.name = $name
            RETURN friend.name AS friend_name
            ORDER BY friend.name
            """,
            name=name
        )

        for record in result:
            print(record["friend_name"])

with GraphDatabase.driver(
    URI,
    auth=(USERNAME, PASSWORD)
) as driver:
    driver.verify_connectivity()

    add_friend(driver, "Arthur", "Guinevere")
    add_friend(driver, "Arthur", "Lancelot")
    add_friend(driver, "Arthur", "Merlin")

    print("Node Person dan relationship KNOWS berhasil dibuat.")
    
    print("\nDaftar teman Arthur:")
    print_friends(driver, "Arthur")
```

jenis graf pertemanan yang telah dibuat di sini merupakan graf sosial (social graph) yang merepresentasikan hubungan pertemanan antarobjek Person. Relasi yag digunakan adalah 'KNOWS' yang menunjukkan bahwa seorang Person mengenal Person lainnya. Pada Contoh tersebut, Arthur memiliki hubungan KNOWS dengan Guinevere, Lancelot, dan Merlin. Relasi tersebut dibuat menggunakan arah panah dari orang yang mengenal menuju orang yang dikenalnya.
