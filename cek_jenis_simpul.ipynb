```python
from getpass import getpass
from neo4j import GraphDatabase
```

```python
URI = "bolt://127.0.0.1:7687"
USERNAME = "neo4j"
PASSWORD = getpass("Masukkan password Neo4j: ")
```

```python
QUERY = """
MATCH (n)
UNWIND labels(n) AS jenis_simpul
RETURN jenis_simpul, COUNT(*) AS jumlah_simpul
ORDER BY jenis_simpul
"""

driver = GraphDatabase.driver(
    URI,
    auth=(USERNAME, PASSWORD)
)

try:
    with driver.session(database="neo4j") as session:
        result = session.run(QUERY)

        print("Jenis simpul dan jumlahnya:")
        for record in result:
            print(
                f"{record['jenis_simpul']}: "
                f"{record['jumlah_simpul']:,}"
            )
finally:
    driver.close()
```
